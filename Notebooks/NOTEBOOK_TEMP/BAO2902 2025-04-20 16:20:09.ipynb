```
MERGE INTO
	dim.PurchaseOrderStatus AS TARGET
USING 
	stg.vw_Dim_PurchaseOrderStatus AS SOURCE

ON 	(TARGET.Code = SOURCE.Code) 

WHEN MATCHED 
	AND (
		IFNULL(TARGET.CODE,'0') <> IFNULL(SOURCE.CODE,'0')
		OR IFNULL(TARGET.DESCRIPTION,'') <> IFNULL(SOURCE.DESCRIPTION,'')
		OR TARGET.IsDeleted = 1
	)
THEN UPDATE 
	SET 
		TARGET.CODE = SOURCE.CODE,
		TARGET.DESCRIPTION = SOURCE.DESCRIPTION,
		TARGET.IsDeleted = 0,
		TARGET.ModDate = CURRENT_TIMESTAMP

WHEN NOT MATCHED THEN 
	INSERT (
		CODE,
		DESCRIPTION,
		CreateDate,
		ModDate
	) 
	VALUES (
		SOURCE.CODE,
		SOURCE.DESCRIPTION,
		CURRENT_TIMESTAMP,
		CURRENT_TIMESTAMP
	)
;
```
